````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# 1. Очищаем названия городов
cities_clean = []
for city in cities:
    cities_clean.append(city.strip().title())
print("Города:", cities_clean)
````

````python
# 2. Переводим суммы из строк в целые числа
amounts_int = []
for amount in amounts:
    amounts_int.append(int(amount))
print("Суммы:", amounts_int)
````

````python
# 3. Число заказов, общая сумма и средний чек
orders_count = len(amounts_int)
all_amounts = sum(amounts_int)
average_amount = round(all_amounts / orders_count)
print("Всего заказов:", orders_count)
print("Общая сумма:", all_amounts, "₸")
print("Средний чек:", average_amount, "₸")
````

````python
# 4. Неоплаченные заказы
unpaid_count = 0
unpaid_total = 0
for amount, is_paid in zip(amounts_int, paid):
    if not is_paid:
        unpaid_count += 1
        unpaid_total += amount
print("Неоплаченных заказов:", unpaid_count)
print("Сумма неоплаченных:", unpaid_total, "₸")
````

````python
# 5. Таблица заказов
print(f"{'Город':<12} | {'Сумма':>8} | Статус")
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    status = "оплачен" if is_paid else "не оплачен"
    print(f"{city:<12} | {amount:>8} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# 6. Оплаченная выручка по Алматы и Астане
almaty_revenue = 0
astana_revenue = 0
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid and city == "Almaty":
        almaty_revenue += amount
    if is_paid and city == "Astana":
        astana_revenue += amount
print("Алматы:", almaty_revenue, "₸")
print("Астана:", astana_revenue, "₸")
````

````python
# 7. Индекс и город самого крупного оплаченного заказа
max_paid_amount = -1
max_paid_index = -1
for i in range(len(amounts_int)):
    if paid[i] and amounts_int[i] > max_paid_amount:
        max_paid_amount = amounts_int[i]
        max_paid_index = i
print("Индекс:", max_paid_index)
print("Город:", cities_clean[max_paid_index])
print("Сумма:", max_paid_amount, "₸")
````

````python
# 8. Доля заказов выходного дня
weekend_count = 0
for day in days:
    if day == "сб" or day == "вс":
        weekend_count += 1
weekend_percent = weekend_count / len(days) * 100
print(f"Заказов в выходные: {weekend_count} из {len(days)} ({weekend_percent:.1f}%)")
````

````python
# 9. Копилка: 15 000 ₸ в месяц
course_price = 400000
savings = 0
months = 0
while savings < course_price:
    savings += 15000
    months += 1
print("Постоянный взнос:", months, "месяцев; накоплено", savings, "₸")

# Вариант: каждый месяц взнос на 1 000 ₸ больше
savings = 0
months = 0
monthly_payment = 15000
while savings < course_price:
    savings += monthly_payment
    months += 1
    monthly_payment += 1000
print("Растущий взнос:", months, "месяцев; накоплено", savings, "₸")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# 10. Два списка с оплаченной выручкой по городам, без словарей
unique_cities = []
revenue_by_city = []
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if not is_paid:
        continue
    if city in unique_cities:
        i = unique_cities.index(city)
        revenue_by_city[i] += amount
    else:
        unique_cities.append(city)
        revenue_by_city.append(amount)

# Сортируем оба списка вместе по убыванию выручки
for i in range(len(revenue_by_city)):
    for j in range(i + 1, len(revenue_by_city)):
        if revenue_by_city[j] > revenue_by_city[i]:
            revenue_by_city[i], revenue_by_city[j] = revenue_by_city[j], revenue_by_city[i]
            unique_cities[i], unique_cities[j] = unique_cities[j], unique_cities[i]

print("Города:", unique_cities)
print("Выручка:", revenue_by_city)
for city, revenue in zip(unique_cities, revenue_by_city):
    print(f"{city}: {revenue} ₸")
````

````python
# 11. Медиана (если количество элементов чётное, берём среднее двух центральных)
sorted_amounts = sorted(amounts_int)
n = len(sorted_amounts)
if n % 2 == 0:
    median = (sorted_amounts[n // 2 - 1] + sorted_amounts[n // 2]) / 2
else:
    median = sorted_amounts[n // 2]
print("Отсортированные суммы:", sorted_amounts)
print("Медиана:", median, "₸")
````

````python
# 12. Текстовая гистограмма оплаченной выручки по дням
unique_days = []
for day in days:
    if day not in unique_days:
        unique_days.append(day)

for day_to_show in unique_days:
    daily_revenue = 0
    for day, amount, is_paid in zip(days, amounts_int, paid):
        if day == day_to_show and is_paid:
            daily_revenue += amount
    blocks = "█" * (daily_revenue // 5000)
    print(f"{day_to_show} | {blocks}  {daily_revenue}")
````

**13. Проблемы качества данных.** Названия городов были записаны в разном регистре и иногда с пробелами по краям, поэтому один город мог выглядеть как несколько разных значений. Суммы хранились в виде строк, и перед вычислениями их пришлось преобразовать в числа. После очистки данных стало возможно корректно группировать заказы и считать выручку.
